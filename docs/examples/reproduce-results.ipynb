# 3. Reproduce the article's results

Recompute Table 1, the endpoint behaviour of the counterdiabatic term, and the scalability summary from the bundled trajectories. No training is required.

````python
import numpy as np

from quantum_pinn.analysis import (
    PAPER_DISTANCES, energy_table, load_scalability, load_trajectory, physics_checks,
)

for row in energy_table():
    spectrum = ", ".join(f"{value:.4f}" for value in row["spectrum"])
    print(
        f"d = {row['distance']:.1f} A | Re E(t=1) = [{spectrum}] | <H_f> = {row['final_energy']:.6f}"
        f" | error = {row['relative_error_percent']:.1e} % | ground population {row['ground_population']:.5f}"
        f" (without CD {row['adiabatic_ground_population']:.3f})"
    )
````

````python
for distance in PAPER_DISTANCES:
    checks = physics_checks(load_trajectory(distance))
    print(distance, {key: f"{value:.2e}" for key, value in checks.items()})
````

````python
rows = load_scalability()
for row in rows:
    print(f"{row['system']:>3} d={row['distance']:.1f} N_F={row['collocation']:>5}  loss={row['total_loss']:.4f}  time={row['training_seconds'] / 3600:.1f} h")
````
